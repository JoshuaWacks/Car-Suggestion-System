```python
import pandas as pd
import os
import numpy as np
import cv2
```

```python
parent_folder = '../../../data/Car Labelling/Standford Dataset Full Label'
```

```python
train_path_0 = os.path.join(parent_folder,'train-00000-of-00002.parquet')
df_train_0 = pd.read_parquet(train_path_0)
```

```python
df_train_0.head()
```

```python
df_train_0.iloc[0].image
```

```python
image_bytes = df_train_0.iloc[0].image['bytes']
image_bytes
```

```python
nparr = np.frombuffer(image_bytes, np.uint8)

# Decode the NumPy array into an OpenCV image (NumPy array representing the image)
# cv2.IMREAD_COLOR loads the image in color (default)
# You can use cv2.IMREAD_GRAYSCALE for grayscale or -1 for unchanged
image = cv2.imdecode(nparr, cv2.IMREAD_COLOR)

# Check if the image was decoded successfully
if image is None:
    print("Could not decode image from bytes. The data might be corrupted or in an unsupported format.")
else:
    # Now 'image' is an OpenCV image (NumPy array) and you can perform further operations
    cv2.imshow("Decoded Image", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```
