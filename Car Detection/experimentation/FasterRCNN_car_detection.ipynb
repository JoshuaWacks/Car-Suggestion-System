```python
%matplotlib inline

import cv2
from matplotlib import pyplot as plt
import torch
import os
import torchvision
from torchvision.models.detection import fasterrcnn_resnet50_fpn_v2, FasterRCNN_ResNet50_FPN_V2_Weights
# from torchvision.models import list_models
from torchvision.io.image import decode_image
from torchvision.utils import draw_bounding_boxes
from torchvision.transforms.functional import to_pil_image


torch.cuda.is_available()
torch.cuda.device_count()
```

```python
weights = FasterRCNN_ResNet50_FPN_V2_Weights.DEFAULT
model = fasterrcnn_resnet50_fpn_v2(weights=weights, box_score_thresh=0.7)
_ = model.eval()
```

```python
# Step 2: Initialize the inference transforms
preprocess = weights.transforms()
```

```python
image_1 = decode_image('../../data/Car Detection/Dataset_2/Data/Data/combined/car0.jpg')
```

```python

# Step 3: Apply inference preprocessing transforms
batch = [preprocess(image_1)]
```

```python
prediction = model(batch)[0]
```

```python
prediction[0]
```

```python
import numpy as np
```

```python
car_preds= []
for box,label in zip(prediction["boxes"],prediction['labels']):
    if label == 3:
        car_preds.append(np.append(box.detach().numpy(), label))

car_preds[0][0:4], car_preds[0][4]
```

```python
labels = [weights.meta["categories"][i] for i in prediction["labels"]]
box = draw_bounding_boxes(image_1, boxes=prediction["boxes"],
                          labels=labels,
                          colors="red",
                          width=4, font_size=30)
im = to_pil_image(box.detach())
im.show()
```
